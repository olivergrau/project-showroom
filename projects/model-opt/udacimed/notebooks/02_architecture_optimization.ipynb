```python
# Analyze the new memory footprint
def count_unique_parameters(model):
    """Count actual unique parameter tensors (shared params counted once)"""
    unique_params = {}
    total_params = 0
    
    for name, param in model.named_parameters():
        param_id = id(param)
        if param_id not in unique_params:
            unique_params[param_id] = {
                'tensor': param,
                'shape': param.shape,
                'elements': param.numel(),
                'first_seen': name,
                'shared_with': []
            }
        else:
            unique_params[param_id]['shared_with'].append(name)
    
    total_unique_elements = sum(info['elements'] for info in unique_params.values())
    
    return total_unique_elements, len(unique_params), unique_params

baseline_unique, baseline_objects, _ = count_unique_parameters(baseline_model)
optimized_unique, optimized_objects, optimized_params_details = count_unique_parameters(optimized_model)

baseline_state_size = sum(p.numel() for p in baseline_model.parameters())
optimized_state_size = sum(p.numel() for p in optimized_model.parameters())

print(f"Baseline model:")
print(f"   State dict size: {baseline_state_size:,} elements")
print(f"   Unique parameters: {baseline_unique:,} elements")

print(f"Optimized model:")
print(f"   State dict size: {optimized_state_size:,} elements")  # Remove the "(same as baseline)" text!
print(f"   Unique parameters: {optimized_unique:,} elements")

reduction = ((baseline_state_size - optimized_state_size) / baseline_state_size * 100)
print(f"Total parameter reduction: {reduction:.1f}%")

n_shared_params = sum([len(param_details['shared_with']) for param_id, param_details in optimized_params_details.items()])
print(f"# Shared parameters: {n_shared_params}")
```

## Step 4: Initialize optimized model via transfer learning

While we have changed the architecture, we can still transfer knowledge from the trained baseline model to our optimized architecture for any matching layers.

Preserving learned features where possible reduces retraining time and maintains clinical performance.

```python
# Transfer compatible weights from baseline to optimized model
# This preserves learned features where possible, reducing retraining time

from typing import Dict, Tuple, List, Any
import torch
import torch.nn as nn


def transfer_compatible_weights(
    baseline_model: nn.Module,
    optimized_model: nn.Module,
    prefer_buffers: bool = True,
    verbose: bool = True,
) -> Dict[str, Any]:
    """
    Transfer all compatible weights/buffers from baseline_model to optimized_model.

    Compatibility rules:
      - If a key exists in both state_dicts and tensor shapes match: transfer.
      - Otherwise try suffix matching (handles wrappers like inner.model.*):
          Match keys by the last N path elements (default strategy is full suffix).
          Only transfer if the suffix match is unique and shapes match.
      - Skip everything else.

    Returns:
      A report dict with counts and a few example keys.
    """
    base_sd = baseline_model.state_dict()
    opt_sd = optimized_model.state_dict()

    # Build reverse index for suffix matching: suffix -> list of baseline keys
    # We also keep shape info for disambiguation.
    suffix_index: Dict[str, List[str]] = {}
    for k in base_sd.keys():
        suffix_index.setdefault(k, []).append(k)  # full key
        # Also index by progressively shorter suffixes for robustness
        parts = k.split(".")
        # We start at 2 to avoid matching only "weight"/"bias"
        for n in range(2, min(len(parts), 6) + 1):
            suf = ".".join(parts[-n:])
            suffix_index.setdefault(suf, []).append(k)

    transferred = []
    skipped = []
    ambiguous = []
    not_found = []

    new_opt_sd = dict(opt_sd)  # copy so we can update entries

    def _is_tensor(x):
        return torch.is_tensor(x)

    # Pass 1: exact matches
    for k_opt, v_opt in opt_sd.items():
        if k_opt in base_sd and _is_tensor(base_sd[k_opt]) and _is_tensor(v_opt):
            v_base = base_sd[k_opt]
            if v_base.shape == v_opt.shape:
                new_opt_sd[k_opt] = v_base
                transferred.append(k_opt)

    # Pass 2: suffix matches for remaining keys
    for k_opt, v_opt in opt_sd.items():
        if k_opt in transferred:
            continue
        if not _is_tensor(v_opt):
            continue

        parts_opt = k_opt.split(".")
        candidates: List[str] = []

        # Try suffix lengths from longest to shorter
        for n in range(min(len(parts_opt), 6), 1, -1):
            suf = ".".join(parts_opt[-n:])
            for cand in suffix_index.get(suf, []):
                if cand in base_sd and _is_tensor(base_sd[cand]) and base_sd[cand].shape == v_opt.shape:
                    candidates.append(cand)
            if candidates:
                break  # found something at this suffix length

        # Deduplicate candidates
        candidates = list(dict.fromkeys(candidates))

        if len(candidates) == 1:
            new_opt_sd[k_opt] = base_sd[candidates[0]]
            transferred.append(k_opt)
        elif len(candidates) == 0:
            not_found.append(k_opt)
        else:
            # More than one possible baseline tensor matches this suffix+shape
            ambiguous.append((k_opt, candidates))
            skipped.append(k_opt)

    # Load the merged state dict
    missing_keys, unexpected_keys = optimized_model.load_state_dict(new_opt_sd, strict=False)

    # Optional: prioritize buffers (BatchNorm running stats) if desired
    # In practice, they are already included in state_dict and handled above.
    # This flag exists to make intent explicit.
    if not prefer_buffers:
        pass

    report = {
        "baseline_tensors": len(base_sd),
        "optimized_tensors": len(opt_sd),
        "transferred": len(transferred),
        "not_found": len(not_found),
        "ambiguous": len(ambiguous),
        "missing_keys_after_load": list(missing_keys),
        "unexpected_keys_after_load": list(unexpected_keys),
        "examples": {
            "transferred": transferred[:10],
            "not_found": not_found[:10],
            "ambiguous": ambiguous[:3],
        },
    }

    if verbose:
        print("TRANSFER COMPATIBLE WEIGHTS REPORT")
        print(f"   Baseline tensors:   {report['baseline_tensors']}")
        print(f"   Optimized tensors:  {report['optimized_tensors']}")
        print(f"   Transferred:        {report['transferred']}")
        print(f"   Not found:          {report['not_found']}")
        print(f"   Ambiguous skipped:  {report['ambiguous']}")
        if report["missing_keys_after_load"]:
            print(f"   Missing after load (expected for changed layers): {len(report['missing_keys_after_load'])}")
        if report["unexpected_keys_after_load"]:
            print(f"   Unexpected after load: {len(report['unexpected_keys_after_load'])}")

        # Show a short hint for grouped conv cases
        if report["missing_keys_after_load"]:
            hint = [k for k in report["missing_keys_after_load"] if "conv" in k and "weight" in k][:5]
            if hint:
                print("   Example missing conv weights (likely due to grouping or other structural changes):")
                for k in hint:
                    print(f"     - {k}")

    return report

print("Transferring compatible weights from baseline to optimized model...")
report = transfer_compatible_weights(baseline_model, optimized_model)

import json

print("\nWeight transfer report:")
print()

# display json of report nicely
print(json.dumps(report, indent=4))
```

### Experimental results

During this notebook, I systematically evaluated multiple architectural optimization techniques, both individually and in combination, to understand their isolated effects and interactions. ### Key insights and learnings

* **Most impactful single technique**
  Interpolation removal was the single most impactful and lowest-risk optimization. It drastically reduced activation memory and improved throughput without affecting model parameters or clinical performance.

* **Most impactful combination**
  The combination of interpolation removal, channel optimization, and aggressive grouped convolution (groups = 8) yielded the best overall result, achieving all production targets simultaneously.

* **Unexpected findings**
  Contrary to initial expectations, increasing the group factor to 8 did not cause a catastrophic drop in sensitivity. This indicates significant redundancy in late-stage ResNet feature channels for this task and resolution.

* **Diminishing returns**
  Moderate grouping (g = 2 or 4) reduced parameters and FLOPs but did not reliably improve latency. Only aggressive grouping crossed the hardware efficiency threshold where latency improvements became measurable.

* **Alignment with Notebook 1 predictions**
  The results strongly aligned with the baseline analysis: late-stage convolutions dominated compute, and architectural changes in those stages provided the highest payoff with minimal clinical risk.

* **Clinical safety considerations**
  Sensitivity remained above 99% across all final configurations. The small drop (-0.3%) relative to baseline is well within acceptable clinical margins and is offset by substantial operational gains.

* **Additional experimentation (given more time)**
  Further experiments could explore:

  * Per-layer group factors instead of a uniform value
  * Sensitivity robustness across multiple operating thresholds
  * Comparison against a lightweight MobileNet-style baseline for reference

### Final optimization strategy

The final architecture adopts the following strategy:

* **Native-resolution inference** to eliminate unnecessary spatial upsampling
* **Aggressive grouped convolution (groups = 8)** applied selectively to late ResNet stages
* **Channels-last memory layout** to improve hardware utilization

This configuration achieves a strong balance between efficiency and safety, avoids unnecessary architectural complexity, and remains close enough to the original ResNet-18 structure to retain interpretability and stability.
